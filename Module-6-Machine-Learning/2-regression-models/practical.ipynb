# Mashq 1: Avtomobil Narxini Bashorat Qilish 🚗

## Vazifa
Avtomobilning turli xususiyatlariga qarab uning narxini bashorat qiling.

## Ma'lumotlar
- `year`: Ishlab chiqarilgan yil
- `km_driven`: Bosib o'tilgan masofa (km)
- `fuel`: Yoqilg'i turi (Petrol/Diesel)
- `seller_type`: Sotuvchi turi
- `transmission`: Uzatma (Manual/Automatic)
- `owner`: Egalar soni
- `mileage`: Yoqilg'i iste'moli (km/l)
- `engine`: Dvigatel hajmi (CC)
- `max_power`: Maksimal quvvat (bhp)
- `seats`: O'rindiqlar soni
- `selling_price`: Sotilish narxi (TARGET)

```python
# Ma'lumotlar yaratish (real holat simulyatsiyasi)
np.random.seed(42)
n = 500

cars_data = pd.DataFrame({
    'year': np.random.randint(2010, 2024, n),
    'km_driven': np.random.randint(5000, 150000, n),
    'fuel': np.random.choice(['Petrol', 'Diesel'], n),
    'transmission': np.random.choice(['Manual', 'Automatic'], n),
    'owner': np.random.choice([0, 1, 2, 3], n, p=[0.5, 0.3, 0.15, 0.05]),
    'mileage': np.random.uniform(10, 30, n),
    'engine': np.random.randint(800, 3000, n),
    'max_power': np.random.uniform(50, 300, n),
    'seats': np.random.choice([2, 4, 5, 7, 8], n, p=[0.05, 0.15, 0.6, 0.15, 0.05])
})

# Narxni hisoblash (murakkab formula)
cars_data['selling_price'] = (
    50000 +
    (cars_data['year'] - 2010) * 15000 -
    cars_data['km_driven'] * 0.5 +
    np.where(cars_data['fuel'] == 'Diesel', 50000, 0) +
    np.where(cars_data['transmission'] == 'Automatic', 100000, 0) -
    cars_data['owner'] * 30000 +
    cars_data['mileage'] * 1000 +
    cars_data['engine'] * 50 +
    cars_data['max_power'] * 500 +
    cars_data['seats'] * 10000 +
    np.random.normal(0, 50000, n)
)

# Manfiy narxlarni to'g'rilash
cars_data['selling_price'] = cars_data['selling_price'].clip(lower=100000)

print("🚗 Avtomobil ma'lumotlari:")
print(cars_data.head(10))
print(f"\nMa'lumotlar shakli: {cars_data.shape}")
```

# Mashq 2: Talaba Baholarini Bashorat Qilish 📚

## Vazifa
Talabaning o'qish vaqti, dars qatnashuvi va boshqa omillarga qarab yakuniy bahosini bashorat qiling.

**Diqqat**: Bu ma'lumotlarda **nochiziqli munosabat** bor!

```python
# Ma'lumotlar yaratish
np.random.seed(42)
n_students = 300

study_data = pd.DataFrame({
    'study_hours': np.random.uniform(0, 10, n_students),
    'attendance': np.random.uniform(40, 100, n_students),
    'previous_score': np.random.uniform(30, 95, n_students),
    'sleep_hours': np.random.uniform(4, 10, n_students),
    'extracurricular': np.random.randint(0, 4, n_students)
})

# Nochiziqli munosabat bilan baho
study_data['final_score'] = (
    30 +
    5 * study_data['study_hours'] +
    0.8 * study_data['study_hours']**2 -  # Quadratic
    0.05 * study_data['study_hours']**3 +  # Cubic
    0.3 * study_data['attendance'] +
    0.2 * study_data['previous_score'] +
    2 * study_data['sleep_hours'] -
    1 * study_data['extracurricular'] +
    np.random.normal(0, 5, n_students)
).clip(0, 100)

print("📚 Talaba ma'lumotlari:")
print(study_data.head(10))
print(f"\nO'rtacha yakuniy baho: {study_data['final_score'].mean():.2f}")
```
