## Mashq 1: Supervised Learning - Klassifikatsiya

**Vazifa:** Iris gullari datasetini ishlating va turlarni klassifikatsiya qiling.

### Bosqichlar:
1. Ma'lumotlarni yuklash
2. Ma'lumotlarni o'rganish (EDA)
3. Train/Test split
4. Model o'qitish (Logistic Regression)
5. Natijalarni baholash

```python
# 1. Ma'lumotlarni yuklash
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target

# DataFrame yaratish (ko'rish uchun qulay)
df = pd.DataFrame(X, columns=iris.feature_names)
df['target'] = y
df['species'] = df['target'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

print("📊 Iris Dataset")
print("="*50)
print(f"Ma'lumotlar: {df.shape}")
print(f"\nBirinchi 5 qator:")
print(df.head())

print(f"\n📈 Statistika:")
print(df.describe())
```

```python
# 2. Ma'lumotlarni vizualizatsiya qilish
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Har bir feature uchun box plot
for idx, col in enumerate(iris.feature_names):
    ax = axes[idx // 2, idx % 2]
    df.boxplot(column=col, by='species', ax=ax)
    ax.set_title(f'{col} by Species')
    ax.set_xlabel('Species')
    ax.set_ylabel(col)

plt.suptitle('Feature Distributions by Species', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```
