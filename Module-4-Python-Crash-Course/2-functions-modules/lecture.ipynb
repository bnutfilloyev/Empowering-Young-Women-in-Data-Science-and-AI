## 6. Funksiyalar yaratish va chaqirish

Funksiya - bu ma'lum vazifani bajaradigan va qayta ishlatilishi mumkin bo'lgan kod bloki.

### Funksiya yaratish sintaksisi:
```python
def funksiya_nomi(parametrlar):
    """Funksiya haqida ma'lumot (docstring)"""
    # Funksiya kodi
    return qiymat  # Ixtiyoriy
```

### Funksiya afzalliklari:
- **Kod qayta ishlatish**: Bir marta yozib, ko'p marta ishlatish
- **Modullik**: Kodni kichik qismlarga bo'lish
- **Oson test qilish**: Alohida funksiyalarni test qilish
- **O'qishga qulay**: Kod tuzilishi aniq bo'ladi

````python
# Funksiyalar yaratish va chaqirish misollar

# 1. Oddiy funksiya (parametrsiz)
def salom_ayt() -> None:
    """Bu funksiya salomlashish uchun ishlatiladi."""
    print("Assalomu alaykum!")

# Funksiyani chaqirish
print("1. Parametrsiz funksiya:")
salom_ayt()

# 2. Parametrli funksiya
def shaxsiy_salom(ism):
    """Shaxsga qaragan salomlashish"""
    print(f"Salom, {ism}!")

print("\n2. Parametrli funksiya:")
shaxsiy_salom("Ahmad")
shaxsiy_salom("Fatima")

# 3. Return bilan funksiya
def yig_indi(a, b):
    """Ikki sonning yig'indisini qaytaradi"""
    natija = a + b
    return natija

print("\n3. Return bilan funksiya:")
summa = yig_indi(5, 3)
print(f"5 + 3 = {summa}")

# 4. Bir nechta qiymat qaytarish
def to_rtburchak_hisoblash(uzunlik: float, kenglik: float) -> tuple:
    """To'rtburchak yuzi va perimetrini hisoblash"""
    yuza = uzunlik * kenglik
    perimetr = 2 * (uzunlik + kenglik)
    return yuza, perimetr

print("\n4. Bir nechta qiymat qaytarish:")
y, p = to_rtburchak_hisoblash(5, 3)
print(f"Yuza: {y}, Perimetr: {p}")

print("\n" + "="*50)
````

## 7. Funksiya parametrlari va qaytarish qiymatlari

Funksiyalar turli xil parametrlar qabul qilishi va har xil tarzda qaytarish qiymatlarini berishi mumkin.

### Parametr turlari:
- **Positional arguments** - joylashuv bo'yicha argumentlar
- **Keyword arguments** - kalit so'z bilan argumentlar  
- **Default parameters** - standart qiymatli parametrlar
- ***args** - o'zgaruvchan miqdordagi argumentlar
- ****kwargs** - kalit so'zli o'zgaruvchan argumentlar

### Return turlari:
- Bitta qiymat qaytarish
- Bir nechta qiymat qaytarish (tuple)
- None qaytarish (return yo'q)
- Shartli qaytarish

````python
# Funksiya parametrlari misollar

# 1. Default parametrlar
def tanishuv(ism, yosh=18, shahar="Toshkent"):
    """Default parametrlar bilan funksiya"""
    print(f"Mening ismim {ism}, {yosh} yoshdaman, {shahar}da yashayman.")

print("1. Default parametrlar:")
tanishuv("Ali")  # Faqat majburiy parametr
tanishuv("Fatima", 22)  # Ikkita parametr
tanishuv("Bobur", 25, "Samarqand")  # Barcha parametrlar

# 2. Keyword arguments
print("\n2. Keyword arguments:")
tanishuv(yosh=30, ism="Zarina", shahar="Buxoro")  # Tartib muhim emas

# 3. *args - o'zgaruvchan argumentlar
def yig_indi_hisoblash(*sonlar):
    """Istalgan miqdordagi sonlar yig'indisini hisoblash"""
    return sum(sonlar)

print("\n3. *args misollar:")
print(f"Yig'indi: {yig_indi_hisoblash(1, 2, 3)}")
print(f"Yig'indi: {yig_indi_hisoblash(1, 2, 3, 4, 5)}")
print(f"Yig'indi: {yig_indi_hisoblash(10, 20)}")

# 4. **kwargs - keyword argumentlar
def profil_yaratish(**ma_lumotlar):
    """Foydalanuvchi profili yaratish"""
    print("Foydalanuvchi profili:")
    for kalit, qiymat in ma_lumotlar.items():
        print(f"  {kalit}: {qiymat}")


print("\n4. **kwargs misollar:")
profil_yaratish(ism="Ahmad", yosh=25, kasb="Dasturchi", shahar="Toshkent")

print("\n" + "="*50)
````

## 8. O'zgaruvchilar ko'lami (Variable Scope)

Scope - bu o'zgaruvchining qaysi joylarda ko'rinadigan va ishlatilishi mumkin bo'lgan hududidir.

### Scope turlari:
- **Local Scope** - Funksiya ichidagi o'zgaruvchilar
- **Global Scope** - Dastur darajasidagi o'zgaruvchilar
- **Enclosing Scope** - Ichki funksiyalar uchun tashqi funksiya scope'i
- **Built-in Scope** - Python o'zining o'zgaruvchilari

### LEGB qoidasi:
Python o'zgaruvchilarni quyidagi tartibda qidiradi:
1. **L**ocal - Joriy funksiya ichida
2. **E**nclosing - Tashqi funksiya ichida  
3. **G**lobal - Modul darajasida
4. **B**uilt-in - Python built-in nomlarda

````python
# Variable Scope misollar

# Global o'zgaruvchi
global_son = 100

def scope_misal():
    """Local va Global scope misollar"""
    # Local o'zgaruvchi
    local_son = 50
    global_son = 200
    print(f"Funksiya ichida - Global: {global_son}, Local: {local_son}")

print("1. Global va Local scope:")
print(f"Funksiya tashqarida - Global: {global_son}")
scope_misal()
print(f"Funksiya tashqarida - Global: {global_son}")

# Global kalit so'zi
counter = 0

def hisoblagich():
    """Global o'zgaruvchini o'zgartirish"""
    global counter
    counter += 1
    print(f"Hisoblagich: {counter}")

print("\n2. Global kalit so'zi:")
hisoblagich()
hisoblagich()
hisoblagich()

# Enclosing scope misoli
def tashqi_funksiya():
    """Enclosing scope misoli"""
    tashqi_o_zgaruvchi = "Tashqaridan"
    
    def ichki_funksiya():
        print(f"Ichki funksiyada: {tashqi_o_zgaruvchi}")
    
    ichki_funksiya()

print("\n3. Enclosing scope:")
tashqi_funksiya()

# Nonlocal kalit so'zi
def hisoblagich_yaratish():
    """Closure va nonlocal misollar"""
    count = 0
    
    def increment():
        nonlocal count
        count += 1
        return count
    
    return increment

print("\n4. Nonlocal va closure:")
my_counter = hisoblagich_yaratish()
print(f"1-chaqirish: {my_counter()}")
print(f"2-chaqirish: {my_counter()}")
print(f"3-chaqirish: {my_counter()}")

print("\n" + "="*50)
````

## 9. Modullar import qilish

Modul - bu Python fayl bo'lib, unda funksiyalar, o'zgaruvchilar va klasslar bo'lishi mumkin.
Modullar kodni tashkil etish va qayta ishlatish uchun ishlatiladi.

### Import usullari:
- `import module_name` - To'liq modul import
- `from module_name import function_name` - Ma'lum funksiyani import
- `from module_name import *` - Barcha narsalarni import (tavsiya etilmaydi)
- `import module_name as alias` - Taxallus bilan import

### Modul qidirish joylari:
1. Joriy katalog
2. PYTHONPATH o'zgaruvchisi  
3. Standart kutubxona katalogi
4. Site-packages katalogi

````python
# Modullar bilan ishlash misollar
# 1. To'liq modul import qilish
print("1. To'liq modul import:")
try:
    import my_module
    print(f"Salomlashish: {my_module.salomlashish('Sardor')}")
    print(f"Hisoblash: {my_module.hisoblash(10, 5, 'ko\'paytirish')}")
    print(f"PI qiymati: {my_module.PI}")
except ImportError:
    print("my_module topilmadi!")

try:
    import qaysidir_folder.my_module
except ImportError:
    print("qaysidir_folder ichidagi my_module topilmadi!")

# 2. Ma'lum funksiyalarni import qilish  
print("\n2. Ma'lum funksiyalarni import:")
try:
    from my_module import salomlashish, hisoblash
    print(f"Salomlashish: {salomlashish('Malika')}")
    print(f"Hisoblash: {hisoblash(20, 4, 'bo\'lish')}")
except ImportError:
    print("Funksiyalarni import qila olmadim!")

# 3. Taxallus bilan import
print("\n3. Taxallus bilan import:")
try:
    import my_module as mm
    print(f"Author: {mm.AUTHOR}")
    print(f"Ayirish: {mm.hisoblash(15, 7, 'ayirish')}")
except ImportError:
    print("Modulni taxallus bilan import qila olmadim!")

print("\n" + "="*50)
````

## 10. Kutubxonalar bilan ishlash

Python da juda ko'p foydali kutubxonalar mavjud. Ular standart (o'rnatilgan) va uchinchi tomon kutubxonalariga bo'linadi.

### Standart kutubxonalar (Built-in):
- **math** - Matematik funksiyalar
- **random** - Tasodifiy sonlar
- **datetime** - Sana va vaqt
- **os** - Operatsion tizim bilan ishlash
- **sys** - Python interpreteri bilan ishlash
- **json** - JSON ma'lumotlari bilan ishlash

### Uchinchi tomon kutubxonalari:
pip bilan o'rnatiladi: `pip install kutubxona_nomi`
- **requests** - HTTP so'rovlar
- **pandas** - Ma'lumotlar tahlili  
- **numpy** - Matematik hisoblashlar
- **matplotlib** - Grafik chizish

````python
# Standart kutubxonalar bilan ishlash

# 1. Math kutubxonasi
import math

print("1. Math kutubxonasi:")
print(f"π qiymati: {math.pi}")
print(f"e qiymati: {math.e}")
print(f"sqrt(16): {math.sqrt(16)}")
print(f"sin(90°): {math.sin(math.radians(90))}")
print(f"log10(100): {math.log10(100)}")
print(f"factorial(5): {math.factorial(5)}")

# 2. Random kutubxonasi
import random

print("\n2. Random kutubxonasi:")
print(f"Tasodifiy son (0-1): {random.random()}")
print(f"Tasodifiy butun son (1-10): {random.randint(1, 10)}")
print(f"Ro'yxatdan tasodifiy element: {random.choice(['qizil', 'yashil', 'ko\'k'])}")

# Tasodifiy ro'yxat aralashtirish
kartalar = ['♠A', '♥K', '♦Q', '♣J']
random.shuffle(kartalar)
print(f"Aralashgan kartalar: {kartalar}")

# 3. Datetime kutubxonasi
from datetime import datetime, date, timedelta
import time

print("\n3. Datetime kutubxonasi:")
hozir = datetime.now()
print(f"Hozirgi vaqt: {hozir}")
print(f"Sana: {hozir.strftime('%d.%m.%Y')}")
print(f"Vaqt: {hozir.strftime('%H:%M:%S')}")

# Sana hisoblash
ertaga = date.today() + timedelta(days=1)
print(f"Ertaga: {ertaga}")

print("\n" + "="*50)
````
