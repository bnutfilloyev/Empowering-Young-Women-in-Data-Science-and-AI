## 4️⃣ SELECT Operatori Asoslari

### 🔍 SELECT - Ma'lumot olish buyrug'i

**SELECT** - SQL da eng ko'p ishlatiladigan buyruq bo'lib, ma'lumotlar bazasidan ma'lumot olish uchun ishlatiladi.

### 📋 SELECT sintaksisi:

```sql
SELECT ustun1, ustun2, ...
FROM jadval_nomi;
```

### 🌟 SELECT ning asosiy ko'rinishlari:

1. **🔍 Barcha ustunlar**: `SELECT * FROM jadval`
2. **📝 Aniq ustunlar**: `SELECT name, age FROM jadval`
3. **🧮 Hisoblangan ustunlar**: `SELECT name, age * 12 AS months FROM jadval`
4. **🔢 Konstantalar**: `SELECT 'Salom', 123, name FROM jadval`

````python
print("🔍 SELECT OPERATORI MISOLLARI")
print("=" * 30)

# 1. Barcha ustunlarni olish
print("\n1️⃣ Barcha ustunlarni ko'rsatish (birinchi 5 ta):")
query1 = "SELECT * FROM employees LIMIT 5"
result1 = pd.read_sql_query(query1, employees_conn)
print(result1.to_string(index=False))

print("\n📝 SQL so'rovi:", query1)
````

````python
# 2. Aniq ustunlarni olish
print("\n2️⃣ Faqat kerakli ustunlar (ism, lavozim, maosh):")
query2 = "SELECT name, position, salary FROM employees LIMIT 8"
result2 = pd.read_sql_query(query2, employees_conn)
print(result2.to_string(index=False))

print("\n📝 SQL so'rovi:", query2)
````

````python
# 3. Hisoblangan ustunlar
print("\n3️⃣ Hisoblangan ustunlar (yillik maosh):")
query3 = """
SELECT 
    name,
    salary,
    salary * 12 AS yearly_salary,
    salary / 1000 AS salary_k
FROM employees 
LIMIT 6
"""
result3 = pd.read_sql_query(query3, employees_conn)
print(result3.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query3.strip())
````

````python
# 4. Konstantalar bilan ishlash
print("\n4️⃣ Konstantalar va funktsiyalar:")
query4 = """
SELECT 
    'Xodim:' AS label,
    name,
    2025 - age AS birth_year,
    'yil' AS year_label
FROM employees 
LIMIT 5
"""
result4 = pd.read_sql_query(query4, employees_conn)
print(result4.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query4.strip())
````

## 5️⃣ WHERE - Ma'lumotlarni Filtrlash

### 🔍 WHERE buyrug'i

**WHERE** buyrug'i ma'lum shartlarga mos keladigan qatorlarni tanlash uchun ishlatiladi.

### 📋 WHERE sintaksisi:

```sql
SELECT ustun1, ustun2
FROM jadval_nomi
WHERE shart;
```

### 🧮 Taqqoslash operatorlari:

| Operator | Ma'nosi | Misol |
|----------|---------|-------|
| `=` | Teng | `age = 25` |
| `!=` yoki `<>` | Teng emas | `age != 25` |
| `>` | Katta | `salary > 1000000` |
| `<` | Kichik | `age < 30` |
| `>=` | Katta yoki teng | `age >= 18` |
| `<=` | Kichik yoki teng | `age <= 65` |

### 🔗 Mantiqiy operatorlar:

- **AND** - Barcha shartlar bajarilishi kerak
- **OR** - Kamida bitta shart bajarilishi kerak
- **NOT** - Shartni inkor qilish

````python
print("🔍 WHERE - FILTRLASH MISOLLARI")
print("=" * 32)

# 1. Oddiy filtrlash
print("\n1️⃣ IT bo'limidagi xodimlar:")
query5 = "SELECT name, position, department FROM employees WHERE department = 'IT'"
result5 = pd.read_sql_query(query5, employees_conn)
print(f"Topildi: {len(result5)} ta xodim")
print(result5.head().to_string(index=False))

print("\n📝 SQL so'rovi:", query5)
````

````python
# 2. Raqamli filtrlash
print("\n2️⃣ Maoshi 1,000,000 so'mdan ko'p bo'lgan xodimlar:")
query6 = "SELECT name, position, salary FROM employees WHERE salary > 1000000 ORDER BY salary DESC"
result6 = pd.read_sql_query(query6, employees_conn)
print(f"Topildi: {len(result6)} ta xodim")
print(result6.head(8).to_string(index=False))

print("\n📝 SQL so'rovi:", query6)
````

````python
# 3. Murakkab shartlar - AND
print("\n3️⃣ IT bo'limidagi yosh xodimlar (30 yoshdan kichik):")

query7 = """
SELECT name, position, department, age, salary 
FROM employees 
WHERE department = 'IT' AND age < 30
"""
result7 = pd.read_sql_query(query7, employees_conn)
print(f"Topildi: {len(result7)} ta xodim")
print(result7.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query7.strip())
````

````python
# 4. Murakkab shartlar - OR
print("\n4️⃣ IT yoki Marketing bo'limidagi xodimlar:")
query8 = """
SELECT name, position, department, salary 
FROM employees 
WHERE department = 'IT' OR department = 'Marketing'
ORDER BY department, salary DESC
"""
result8 = pd.read_sql_query(query8, employees_conn)
print(f"Topildi: {len(result8)} ta xodim")
print(result8.head(10).to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query8.strip())
````

````python
# 5. IN operatori - bir nechta qiymat
print("\n5️⃣ IN operatori - bir nechta bo'limni tanlash:")
query9 = """
SELECT name, position, department, salary 
FROM employees 
WHERE department IN ('IT', 'Marketing', 'Sales')
ORDER BY department, salary DESC
"""
result9 = pd.read_sql_query(query9, employees_conn)
print(f"Topildi: {len(result9)} ta xodim")
print(result9.head(8).to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query9.strip())
````

````python
# 6. BETWEEN operatori - oralig'ini tanlash
print("\n6️⃣ BETWEEN - yosh oralig'i (25-35):")

query10 = """
SELECT name, age, position, salary 
FROM employees 
WHERE age BETWEEN 25 AND 35
ORDER BY age
"""
result10 = pd.read_sql_query(query10, employees_conn)
print(f"Topildi: {len(result10)} ta xodim")
print(result10.head(8).to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query10.strip())
````

## 6️⃣ ORDER BY - Ma'lumotlarni Tartiblash

### 📊 ORDER BY buyrug'i

**ORDER BY** natijalarni ma'lum ustun bo'yicha tartiblash uchun ishlatiladi.

### 📋 ORDER BY sintaksisi:

```sql
SELECT ustun1, ustun2
FROM jadval_nomi
ORDER BY ustun1 [ASC|DESC];
```

### 🔄 Tartiblash yo'nalishlari:

- **ASC** (Ascending) - O'sish tartibida (default)
- **DESC** (Descending) - Kamayish tartibida

### 📊 Ko'p ustun bo'yicha tartiblash:

```sql
ORDER BY ustun1 DESC, ustun2 ASC
```

````python
print("📊 ORDER BY - TARTIBLASH MISOLLARI")
print("=" * 34)

# 1. Maosh bo'yicha tartiblash (yuqoridan pastga)
print("\n1️⃣ Eng yuqori maoshli xodimlar:")
query11 = """
SELECT name, position, salary 
FROM employees 
ORDER BY salary DESC
LIMIT 8
"""
result11 = pd.read_sql_query(query11, employees_conn)
print(result11.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query11.strip())
````

````python
# 2. Yosh bo'yicha tartiblash (pastdan yuqoriga)
print("\n2️⃣ Eng yosh xodimlar:")
query12 = """
SELECT name, age, position, department 
FROM employees 
ORDER BY age ASC
LIMIT 8
"""
result12 = pd.read_sql_query(query12, employees_conn)
print(result12.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query12.strip())
````

````python
# 3. Ko'p ustun bo'yicha tartiblash
print("\n3️⃣ Bo'lim va maosh bo'yicha tartiblash:")
query13 = """
SELECT name, department, position, salary 
FROM employees 
ORDER BY department ASC, salary DESC
LIMIT 12
"""
result13 = pd.read_sql_query(query13, employees_conn)
print(result13.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query13.strip())
````

````python
# 4. Ism bo'yicha alifbo tartibida
print("\n4️⃣ Alifbo tartibida xodimlar:")
query14 = """
SELECT name, position, department 
FROM employees 
ORDER BY name ASC
LIMIT 8
"""
result14 = pd.read_sql_query(query14, employees_conn)
print(result14.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query14.strip())
````

## 7️⃣ LIMIT - Natijalar Sonini Cheklash

### 🔢 LIMIT buyrug'i

**LIMIT** so'rov natijasida qaytariladigan qatorlar sonini cheklash uchun ishlatiladi.

### 📋 LIMIT sintaksisi:

```sql
SELECT ustun1, ustun2
FROM jadval_nomi
LIMIT son;
```

### 📄 OFFSET bilan sahifalash:

```sql
SELECT ustun1, ustun2
FROM jadval_nomi
LIMIT son OFFSET boshlang'ich_nuqta;
```

### 💡 LIMIT nima uchun muhim?

1. **⚡ Tezlik** - Katta ma'lumotlar bazalarida
2. **💻 Xotira** - Kam resurslarda ishlash
3. **📱 UI/UX** - Sahifalash (pagination)
4. **🔍 Test** - Ma'lumotlarni tekshirish

````python
print("🔢 LIMIT - NATIJALARNI CHEKLASH")
print("=" * 32)

# 1. Eng yuqori 5 ta maosh
print("\n1️⃣ Top 5 yuqori maoshli xodimlar:")
query15 = """
SELECT name, position, salary 
FROM employees 
ORDER BY salary DESC
LIMIT 5
"""
result15 = pd.read_sql_query(query15, employees_conn)
print(result15.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query15.strip())
````

````python
# 2. OFFSET bilan sahifalash
print("\n2️⃣ 6-10 o'rinlardagi yuqori maoshli xodimlar:")
query16 = """
SELECT name, position, salary 
FROM employees 
ORDER BY salary DESC
LIMIT 5 OFFSET 5
"""
result16 = pd.read_sql_query(query16, employees_conn)
print(result16.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query16.strip())
````

````python
# 3. Har bir bo'limdan bitta xodim
print("\n3️⃣ Har bir bo'limdan eng yuqori maoshli xodim:")

# Avval bo'limlarni ko'ramiz
departments_query = "SELECT DISTINCT department FROM employees ORDER BY department"
departments = pd.read_sql_query(departments_query, employees_conn)
print("Bo'limlar:", list(departments['department']))

# Har bir bo'limdan top xodim
query17 = """
SELECT 
    department,
    name,
    position,
    salary,
    RANK() OVER (PARTITION BY department ORDER BY salary DESC) as rank
FROM employees
"""

# SQLite RANK() ni qo'llab-quvvatlamaydi, shuning uchun alternativ yondashuv
query17_alt = """
SELECT DISTINCT
    department,
    (SELECT name FROM employees e2 
     WHERE e2.department = e1.department 
     ORDER BY salary DESC LIMIT 1) as top_employee,
    (SELECT MAX(salary) FROM employees e3 
     WHERE e3.department = e1.department) as max_salary
FROM employees e1
ORDER BY max_salary DESC
"""

result17 = pd.read_sql_query(query17_alt, employees_conn)
print(result17.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query17_alt.strip())
````

## 8️⃣ Murakkab SELECT So'rovlari

### 🔍 Barcha operatorlarni birlashtirish

Haqiqiy loyihalarda odatda bir nechta operatorni birga ishlatish kerak bo'ladi:

```sql
SELECT ustun1, ustun2, hisoblash
FROM jadval_nomi
WHERE shartlar
ORDER BY tartiblash
LIMIT natijalar_soni;
```

### 🎯 Murakkab misollar:

1. **Filtrlash + Tartiblash + Cheklash**
2. **Hisoblangan ustunlar + Shartlar**
3. **Ko'p shartli filtrlash**
4. **String operatorlari (LIKE, UPPER, LOWER)**

````python
print("🔍 MURAKKAB SQL SO'ROVLARI")
print("=" * 28)

# 1. Eng murakkab so'rov - barcha operatorlar
print("\n1️⃣ IT bo'limidagi yosh, yuqori maoshli xodimlar:")
query18 = """
SELECT 
    name,
    position,
    age,
    salary,
    salary * 12 AS yearly_salary,
    CASE 
        WHEN age < 25 THEN 'Yosh'
        WHEN age < 35 THEN 'O`rtacha'
        ELSE 'Katta'
    END AS age_group
FROM employees 
WHERE 
    department = 'IT' 
    AND salary > 800000 
    AND age < 40
ORDER BY salary DESC, age ASC
LIMIT 8
"""

result18 = pd.read_sql_query(query18, employees_conn)
print(result18.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query18.strip())
````

````python
# 2. String operatorlari - LIKE
print("\n2️⃣ Ismida 'a' harfi bo'lgan xodimlar:")
query19 = """
SELECT name, position, department
FROM employees 
WHERE name LIKE '%a%'
ORDER BY name
LIMIT 10
"""
result19 = pd.read_sql_query(query19, employees_conn)
print(f"Topildi: {len(result19)} ta xodim")
print(result19.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query19.strip())
````

````python
# 3. Statistik ma'lumotlar
print("\n3️⃣ Har bir bo'lim bo'yicha statistika:")
query20 = """
SELECT 
    department,
    COUNT(*) as xodimlar_soni,
    AVG(salary) as o_rtacha_maosh,
    MIN(salary) as min_maosh,
    MAX(salary) as max_maosh,
    AVG(age) as o_rtacha_yosh
FROM employees 
GROUP BY department
ORDER BY o_rtacha_maosh DESC
"""
result20 = pd.read_sql_query(query20, employees_conn)

# Sonlarni formatlash
result20['o_rtacha_maosh'] = result20['o_rtacha_maosh'].round(0).astype(int)
result20['o_rtacha_yosh'] = result20['o_rtacha_yosh'].round(1)

print(result20.to_string(index=False))

print("\n📝 SQL so'rovi:")
print(query20.strip())
````

````python
# 4. Ma'lumotlarni vizualizatsiya qilish
print("\n📊 MA'LUMOTLARNI VIZUALIZATSIYA")
print("=" * 32)

# Bo'limlar bo'yicha xodimlar soni
dept_stats = pd.read_sql_query("""
SELECT department, COUNT(*) as count, AVG(salary) as avg_salary
FROM employees 
GROUP BY department
ORDER BY count DESC
""", employees_conn)

# Grafik chizish
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Bo'limlar bo'yicha xodimlar soni
ax1.bar(dept_stats['department'], dept_stats['count'], color='skyblue', alpha=0.8)
ax1.set_title('Bo\'limlar bo\'yicha xodimlar soni', fontsize=14, fontweight='bold')
ax1.set_xlabel('Bo\'lim')
ax1.set_ylabel('Xodimlar soni')
ax1.tick_params(axis='x', rotation=45)
ax1.grid(True, alpha=0.3)

# Bo'limlar bo'yicha o'rtacha maosh
ax2.bar(dept_stats['department'], dept_stats['avg_salary'], color='lightcoral', alpha=0.8)
ax2.set_title('Bo\'limlar bo\'yicha o\'rtacha maosh', fontsize=14, fontweight='bold')
ax2.set_xlabel('Bo\'lim')
ax2.set_ylabel('O\'rtacha maosh (so\'m)')
ax2.tick_params(axis='x', rotation=45)
ax2.grid(True, alpha=0.3)
ax2.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x/1000:.0f}K'))

plt.tight_layout()
plt.show()

print("📊 Grafiklar muvaffaqiyatli yaratildi!")
````

## 9️⃣ Amaliy Mashq - Mahsulotlar Ma'lumotlar Bazasi

### 🛍️ Mahsulotlar bilan ishlash

Endi boshqa ma'lumotlar bazasi bilan ishlashni mashq qilamiz - mahsulotlar katalogi.

````python
print("🛍️ MAHSULOTLAR MA'LUMOTLAR BAZASI")
print("=" * 35)

# Ma'lumotlar bazasi tuzilishi
print("\n📊 PRODUCTS jadvali tuzilishi:")
products_schema = pd.read_sql_query(
    "PRAGMA table_info(products)", 
    products_conn
)
print(products_schema.to_string(index=False))

# Jami mahsulotlar soni
products_count = pd.read_sql_query(
    "SELECT COUNT(*) as total_products FROM products", 
    products_conn
)
print(f"\n🛍️ Jami mahsulotlar: {products_count['total_products'].iloc[0]} ta")
````

````python
# Mahsulotlar bilan ishlash misollari
print("\n1️⃣ Eng qimmat 10 ta mahsulot:")
expensive_products = pd.read_sql_query("""
SELECT name, category, price, rating
FROM products
ORDER BY price DESC
LIMIT 10
""", products_conn)
print(expensive_products.to_string(index=False))

print("\n2️⃣ Elektronika kategoriyasidagi mahsulotlar:")
electronics = pd.read_sql_query("""
SELECT name, price, rating, supplier
FROM products
WHERE category = 'Elektronika'
ORDER BY rating DESC
LIMIT 8
""", products_conn)
print(electronics.to_string(index=False))

print("\n3️⃣ Yuqori reytingli (4.5+), arzon mahsulotlar (<100):")
quality_affordable = pd.read_sql_query("""
SELECT name, category, price, rating, stock
FROM products
WHERE rating >= 4.5 AND price < 100 AND stock > 0
ORDER BY rating DESC, price ASC
LIMIT 10
""", products_conn)
print(quality_affordable.to_string(index=False))
````

````python
# Kategoriyalar bo'yicha statistika
print("\n📊 Kategoriyalar bo'yicha statistika:")
category_stats = pd.read_sql_query("""
SELECT 
    category,
    COUNT(*) as mahsulotlar_soni,
    AVG(price) as o_rtacha_narx,
    AVG(rating) as o_rtacha_reyting,
    SUM(stock) as jami_zaxira
FROM products
GROUP BY category
ORDER BY mahsulotlar_soni DESC
""", products_conn)

# Sonlarni formatlash
category_stats['o_rtacha_narx'] = category_stats['o_rtacha_narx'].round(2)
category_stats['o_rtacha_reyting'] = category_stats['o_rtacha_reyting'].round(2)

print(category_stats.to_string(index=False))

# Vizualizatsiya
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Kategoriyalar bo'yicha mahsulotlar soni
axes[0,0].bar(category_stats['category'], category_stats['mahsulotlar_soni'], 
              color='lightblue', alpha=0.8)
axes[0,0].set_title('Kategoriyalar bo\'yicha mahsulotlar soni')
axes[0,0].tick_params(axis='x', rotation=45)
axes[0,0].grid(True, alpha=0.3)

# 2. O'rtacha narx
axes[0,1].bar(category_stats['category'], category_stats['o_rtacha_narx'], 
              color='lightgreen', alpha=0.8)
axes[0,1].set_title('Kategoriyalar bo\'yicha o\'rtacha narx')
axes[0,1].tick_params(axis='x', rotation=45)
axes[0,1].grid(True, alpha=0.3)

# 3. O'rtacha reyting
axes[1,0].bar(category_stats['category'], category_stats['o_rtacha_reyting'], 
              color='orange', alpha=0.8)
axes[1,0].set_title('Kategoriyalar bo\'yicha o\'rtacha reyting')
axes[1,0].tick_params(axis='x', rotation=45)
axes[1,0].grid(True, alpha=0.3)
axes[1,0].set_ylim(0, 5)

# 4. Jami zaxira
axes[1,1].bar(category_stats['category'], category_stats['jami_zaxira'], 
              color='pink', alpha=0.8)
axes[1,1].set_title('Kategoriyalar bo\'yicha jami zaxira')
axes[1,1].tick_params(axis='x', rotation=45)
axes[1,1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📊 Mahsulotlar statistikasi grafiklar bilan taqdim etildi!")
````
