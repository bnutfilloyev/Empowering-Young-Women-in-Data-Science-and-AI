# 🔬 Amaliy Mashg'ulot: SQL SELECT Asoslari

## 🎯 Mashg'ulot maqsadi
SQL SELECT operatori va uning asosiy komponentlari (WHERE, ORDER BY, LIMIT) bilan amaliy ishlash ko'nikmalarini rivojlantirish.

```python
# Kutubxonalar
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Ma'lumotlar bazasiga ulanish
employees_conn = sqlite3.connect('datasets/employees.db')
products_conn = sqlite3.connect('datasets/products.db')
students_conn = sqlite3.connect('datasets/students.db')
```

## 1️⃣ Ma'lumotlar Bazasi bilan Tanishish

### 📊 Bizda 3 ta ma'lumotlar bazasi mavjud:
- **employees.db** - Xodimlar ma'lumotlari
- **products.db** - Mahsulotlar katalogi
- **students.db** - Talabalar reytingi

```python
# VAZIFA 1: Har bir ma'lumotlar bazasining tuzilishini o'rganing

print("📊 MA'LUMOTLAR BAZASI TUZILISHI")
print("=" * 33)

# 1.1 Employees jadval tuzilishi
print("\n🏢 EMPLOYEES (Xodimlar):")
employees_info = pd.read_sql_query(
    "SELECT * FROM employees LIMIT 3", 
    employees_conn
)
print(employees_info.to_string(index=False))
print(f"Jami yozuvlar: {pd.read_sql_query('SELECT COUNT(*) as count FROM employees', employees_conn)['count'][0]}")

# 1.2 Products jadval tuzilishi
print("\n🛍️ PRODUCTS (Mahsulotlar):")
products_info = pd.read_sql_query(
    "SELECT * FROM products LIMIT 3", 
    products_conn
)
print(products_info.to_string(index=False))
print(f"Jami yozuvlar: {pd.read_sql_query('SELECT COUNT(*) as count FROM products', products_conn)['count'][0]}")

# 1.3 Students jadval tuzilishi
print("\n🎓 STUDENTS (Talabalar):")
students_info = pd.read_sql_query(
    "SELECT * FROM students LIMIT 3", 
    students_conn
)
print(students_info.to_string(index=False))
print(f"Jami yozuvlar: {pd.read_sql_query('SELECT COUNT(*) as count FROM students', students_conn)['count'][0]}")
```

## 2️⃣ Oddiy SELECT So'rovlari

### 🎯 Maqsad: SELECT ning turli ko'rinishlarini mashq qilish

#### 📝 Sizning vazifangiz:
Quyidagi so'rovlarni yozing va natijalarni tahlil qiling.

```python
# VAZIFA 2.1: Employees jadvalidan faqat ism va lavozimni oling

print("📝 VAZIFA 2.1: Ism va lavozim")
print("=" * 28)

# Sizning kodingiz:
query_2_1 = """
-- Bu yerga o'z kodingizni yozing
SELECT name, position 
FROM employees 
LIMIT 10
"""

result_2_1 = pd.read_sql_query(query_2_1, employees_conn)
print(result_2_1.to_string(index=False))

print(f"\n✅ Natija: {len(result_2_1)} ta xodim")
```

```python
# VAZIFA 2.2: Products jadvalidan mahsulot nomi, narxi va reytingini oling

print("📝 VAZIFA 2.2: Mahsulot ma'lumotlari")
print("=" * 32)

# Sizning kodingiz (bu yerga yozing):
query_2_2 = """
SELECT name, price, rating
FROM products
LIMIT 8
"""

result_2_2 = pd.read_sql_query(query_2_2, products_conn)
print(result_2_2.to_string(index=False))

print(f"\n✅ Natija: {len(result_2_2)} ta mahsulot")
```

```python
# VAZIFA 2.3: Hisoblangan ustunlar - yillik maosh

print("📝 VAZIFA 2.3: Yillik maosh hisoblash")
print("=" * 33)

# Sizning kodingiz:
query_2_3 = """
SELECT 
    name,
    salary,
    salary * 12 AS yearly_salary,
    department
FROM employees
LIMIT 8
"""

result_2_3 = pd.read_sql_query(query_2_3, employees_conn)
print(result_2_3.to_string(index=False))

print(f"\n✅ Yillik maosh hisoblab topildi")
```

```python
# VAZIFA 2.4: Mustaqil mashq - Students jadvali bilan ishlang

print("📝 VAZIFA 2.4: Talabalar GPA ma'lumoti")
print("=" * 34)

# TOPSHIRIQ: Talaba ismi, fakultet, kursi va GPA ni ko'rsating
# Qo'shimcha: GPA ni 100 ga ko'paytiring (foiz ko'rinishida)

query_2_4 = """
-- Bu yerga o'z kodingizni yozing
SELECT 
    name,
    faculty,
    course,
    gpa,
    gpa * 25 AS gpa_percentage
FROM students
LIMIT 10
"""

result_2_4 = pd.read_sql_query(query_2_4, students_conn)
print(result_2_4.to_string(index=False))

print(f"\n✅ {len(result_2_4)} ta talaba ma'lumoti")
```

## 3️⃣ WHERE bilan Filtrlash

### 🎯 Maqsad: Turli filtrlash usullarini o'rganish

#### 📝 Quyidagi vazifalarni bajaring:

```python
# VAZIFA 3.1: IT bo'limidagi xodimlarni toping

print("📝 VAZIFA 3.1: IT xodimlari")
print("=" * 23)

query_3_1 = """
SELECT name, position, department, salary
FROM employees
WHERE department = 'IT'
ORDER BY salary DESC
"""

result_3_1 = pd.read_sql_query(query_3_1, employees_conn)
print(result_3_1.to_string(index=False))

print(f"\n✅ IT bo'limida {len(result_3_1)} ta xodim ishlaydi")
```

```python
# VAZIFA 3.2: Maoshi 1,200,000 so'mdan ko'p bo'lgan xodimlar

print("📝 VAZIFA 3.2: Yuqori maoshli xodimlar")
print("=" * 33)

# Sizning kodingiz:
query_3_2 = """
SELECT name, position, salary, department
FROM employees
WHERE salary > 1200000
ORDER BY salary DESC
"""

result_3_2 = pd.read_sql_query(query_3_2, employees_conn)
print(result_3_2.to_string(index=False))

print(f"\n✅ {len(result_3_2)} ta xodim yuqori maosh oladi")
print(f"📊 O'rtacha maosh: {result_3_2['salary'].mean():,.0f} so'm")
```

```python
# VAZIFA 3.3: Murakkab shart - yosh va maosh

print("📝 VAZIFA 3.3: Yosh va yuqori maoshli xodimlar")
print("=" * 39)

# TOPSHIRIQ: 30 yoshdan kichik VA maoshi 800,000 dan ko'p bo'lgan xodimlar

query_3_3 = """
-- Bu yerga o'z kodingizni yozing
SELECT name, age, position, salary, department
FROM employees
WHERE age < 30 AND salary > 800000
ORDER BY age, salary DESC
"""

result_3_3 = pd.read_sql_query(query_3_3, employees_conn)
print(result_3_3.to_string(index=False))

print(f"\n✅ {len(result_3_3)} ta yosh, yuqori maoshli xodim")
if len(result_3_3) > 0:
    print(f"📊 O'rtacha yosh: {result_3_3['age'].mean():.1f}")
    print(f"📊 O'rtacha maosh: {result_3_3['salary'].mean():,.0f} so'm")
```

```python
# VAZIFA 3.4: IN operatori - bir nechta qiymat

print("📝 VAZIFA 3.4: Ko'p bo'limli so'rov")
print("=" * 30)

# TOPSHIRIQ: IT, Marketing va Sales bo'limlaridagi xodimlar

query_3_4 = """
SELECT department, name, position, salary
FROM employees
WHERE department IN ('IT', 'Marketing', 'Sales')
ORDER BY department, salary DESC
"""

result_3_4 = pd.read_sql_query(query_3_4, employees_conn)
print(result_3_4.head(15).to_string(index=False))

print(f"\n✅ Jami: {len(result_3_4)} ta xodim")

# Bo'limlar bo'yicha taqsimot
dept_count = result_3_4['department'].value_counts()
print("\n📊 Bo'limlar bo'yicha taqsimot:")
for dept, count in dept_count.items():
    print(f"   {dept}: {count} ta")
```

```python
# VAZIFA 3.5: Mahsulotlar bilan ishlash

print("📝 VAZIFA 3.5: Elektronika mahsulotlari")
print("=" * 33)

# TOPSHIRIQ: Elektronika kategoriyasidagi, narxi 100-500 oralig'idagi mahsulotlar

query_3_5 = """
-- Bu yerga o'z kodingizni yozing
SELECT name, price, rating, stock, supplier
FROM products
WHERE category = 'Elektronika' AND price BETWEEN 100 AND 500
ORDER BY rating DESC, price ASC
"""

result_3_5 = pd.read_sql_query(query_3_5, products_conn)
print(result_3_5.to_string(index=False))

print(f"\n✅ {len(result_3_5)} ta elektronika mahsuloti topildi")
if len(result_3_5) > 0:
    print(f"📊 O'rtacha narx: {result_3_5['price'].mean():.2f}")
    print(f"📊 O'rtacha reyting: {result_3_5['rating'].mean():.2f}")
```

## 4️⃣ ORDER BY va LIMIT

### 🎯 Maqsad: Ma'lumotlarni tartiblash va natija hajmini boshqarish

```python
# VAZIFA 4.1: Eng yuqori GPA ga ega talabalar

print("📝 VAZIFA 4.1: Top 10 talaba")
print("=" * 25)

query_4_1 = """
SELECT name, faculty, course, gpa, city
FROM students
ORDER BY gpa DESC
LIMIT 10
"""

result_4_1 = pd.read_sql_query(query_4_1, students_conn)
print(result_4_1.to_string(index=False))

print(f"\n✅ Eng yuqori GPA: {result_4_1['gpa'].max()}")
print(f"📊 Top 10 o'rtacha GPA: {result_4_1['gpa'].mean():.3f}")
```

```python
# VAZIFA 4.2: Sahifalash (Pagination) - OFFSET bilan

print("📝 VAZIFA 4.2: Sahifalash misoli")
print("=" * 29)

# 1-sahifa: Birinchi 5 ta eng qimmat mahsulot
print("\n📄 1-SAHIFA (Top 1-5):")
query_4_2a = """
SELECT name, category, price, rating
FROM products
ORDER BY price DESC
LIMIT 5
"""
result_4_2a = pd.read_sql_query(query_4_2a, products_conn)
print(result_4_2a.to_string(index=False))

# 2-sahifa: 6-10 o'rinlardagi mahsulotlar
print("\n📄 2-SAHIFA (Top 6-10):")
query_4_2b = """
-- Bu yerga OFFSET bilan kod yozing
SELECT name, category, price, rating
FROM products
ORDER BY price DESC
LIMIT 5 OFFSET 5
"""
result_4_2b = pd.read_sql_query(query_4_2b, products_conn)
print(result_4_2b.to_string(index=False))
```

```python
# VAZIFA 4.3: Ko'p ustun bo'yicha tartiblash

print("📝 VAZIFA 4.3: Murakkab tartiblash")
print("=" * 31)

# TOPSHIRIQ: Avval fakultet bo'yicha, keyin GPA bo'yicha tartiblang

query_4_3 = """
-- Bu yerga kod yozing
SELECT faculty, name, course, gpa
FROM students
ORDER BY faculty ASC, gpa DESC
"""

result_4_3 = pd.read_sql_query(query_4_3, students_conn)
print(result_4_3.head(15).to_string(index=False))

print(f"\n✅ {len(result_4_3)} ta talaba fakultet va GPA bo'yicha tartiblandi")
```

## 5️⃣ Murakkab So'rovlar

### 🎯 Maqsad: Barcha o'rgangan operatorlarni birlashtirib murakkab so'rovlar yaratish

```python
# VAZIFA 5.1: Eng murakkab so'rov - xodimlar uchun

print("📝 VAZIFA 5.1: Super murakkab so'rov")
print("=" * 32)

# TOPSHIRIQ: IT/Marketing bo'limidagi, yoshi 25-40 oralig'idagi, 
# maoshi 900k dan ko'p bo'lgan xodimlarni maosh bo'yicha tartiblang

query_5_1 = """
SELECT 
    name,
    position,
    department,
    age,
    salary,
    salary * 12 AS yearly_salary,
    CASE 
        WHEN salary >= 1500000 THEN 'Yuqori'
        WHEN salary >= 1000000 THEN 'O\'rtacha'
        ELSE 'Past'
    END AS salary_level
FROM employees
WHERE 
    department IN ('IT', 'Marketing') 
    AND age BETWEEN 25 AND 40
    AND salary > 900000
ORDER BY salary DESC, age ASC
LIMIT 8
"""

result_5_1 = pd.read_sql_query(query_5_1, employees_conn)
print(result_5_1.to_string(index=False))

print(f"\n✅ {len(result_5_1)} ta xodim shartlarga javob berdi")
if len(result_5_1) > 0:
    print(f"📊 O'rtacha yosh: {result_5_1['age'].mean():.1f}")
    print(f"📊 O'rtacha maosh: {result_5_1['salary'].mean():,.0f} so'm")
    print(f"📊 O'rtacha yillik: {result_5_1['yearly_salary'].mean():,.0f} so'm")
```

```python
# VAZIFA 5.2: Mahsulotlar uchun murakkab so'rov

print("📝 VAZIFA 5.2: Premium mahsulotlar")
print("=" * 30)

# TOPSHIRIQ: Reytingi 4.0 dan yuqori, zaxirasi mavjud bo'lgan, 
# narxi 50-300 oralig'idagi mahsulotlar

query_5_2 = """
-- Bu yerga murakkab so'rov yozing
SELECT 
    name,
    category,
    price,
    rating,
    stock,
    supplier,
    CASE 
        WHEN rating >= 4.5 THEN 'Excellent'
        WHEN rating >= 4.0 THEN 'Good'
        ELSE 'Average'
    END AS quality_level
FROM products
WHERE 
    rating > 4.0 
    AND stock > 0 
    AND price BETWEEN 50 AND 300
ORDER BY rating DESC, price ASC
LIMIT 10
"""

result_5_2 = pd.read_sql_query(query_5_2, products_conn)
print(result_5_2.to_string(index=False))

print(f"\n✅ {len(result_5_2)} ta premium mahsulot")
if len(result_5_2) > 0:
    print(f"📊 O'rtacha reyting: {result_5_2['rating'].mean():.2f}")
    print(f"📊 O'rtacha narx: {result_5_2['price'].mean():.2f}")
    print(f"📊 Jami zaxira: {result_5_2['stock'].sum()} ta")
```

```python
# VAZIFA 5.3: String operatorlari - LIKE

print("📝 VAZIFA 5.3: Ism bo'yicha qidiruv")
print("=" * 31)

# TOPSHIRIQ: Ismida 'a' harfi bo'lgan, stipendiya olayotgan talabalar

query_5_3 = """
SELECT 
    name,
    faculty,
    gpa,
    scholarship,
    city
FROM students
WHERE 
    name LIKE '%a%' 
    AND scholarship = 1
ORDER BY gpa DESC
"""

result_5_3 = pd.read_sql_query(query_5_3, students_conn)
print(result_5_3.to_string(index=False))

print(f"\n✅ {len(result_5_3)} ta stipendiyachi talaba ('a' harfi bilan)")
if len(result_5_3) > 0:
    print(f"📊 O'rtacha GPA: {result_5_3['gpa'].mean():.3f}")
    print(f"📊 Eng yuqori GPA: {result_5_3['gpa'].max():.3f}")
```

```python
# VAZIFA 5.4: Mustaqil loyiha - o'zingiz so'rov yarating

print("📝 VAZIFA 5.4: MUSTAQIL LOYIHA")
print("=" * 29)
print("🎯 Vazifa: O'zingiz uchun qiziqarli so'rov yarating")

# TOPSHIRIQ: 
# 1. Biror mavzu tanlang (masalan: "Eng yaxshi IT xodimlari")
# 2. Murakkab shart yarating (WHERE bilan)
# 3. Tartiblab chiqaring (ORDER BY)
# 4. Natijani cheklang (LIMIT)
# 5. Hisoblangan ustun qo'shing

# Misol: "Toshkentdagi eng yaxshi IT talabalar"
my_query = """
-- Bu yerga o'zingizning ijodiy so'rovingizni yozing
SELECT 
    name,
    faculty,
    course,
    gpa,
    city,
    gpa * 25 AS percentage_score,
    CASE 
        WHEN gpa >= 3.8 THEN 'A+'
        WHEN gpa >= 3.5 THEN 'A'
        WHEN gpa >= 3.0 THEN 'B'
        ELSE 'C'
    END AS grade
FROM students
WHERE 
    city = 'Toshkent' 
    AND faculty IN ('Informatika', 'Muhandislik')
    AND gpa >= 3.0
ORDER BY gpa DESC, course ASC
LIMIT 8
"""

my_result = pd.read_sql_query(my_query, students_conn)
print("\n🏆 SIZNING NATIJANGIZ:")
print(my_result.to_string(index=False))

print(f"\n✅ {len(my_result)} ta natija topildi")
print("\n💡 Tahlil:")
if len(my_result) > 0:
    print(f"   📊 O'rtacha GPA: {my_result['gpa'].mean():.3f}")
    print(f"   📊 Eng yuqori GPA: {my_result['gpa'].max():.3f}")
    print(f"   📊 O'rtacha foiz: {my_result['percentage_score'].mean():.1f}%")
    
    # Grade taqsimoti
    grade_counts = my_result['grade'].value_counts()
    print("   📊 Grade taqsimoti:")
    for grade, count in grade_counts.items():
        print(f"      {grade}: {count} ta")
```

## 6️⃣ Natijalarni Vizualizatsiya

### 📊 SQL natijalarini grafik ko'rinishida taqdim etish

```python
# Ma'lumotlarni vizualizatsiya qilish
print("📊 NATIJALARNI VIZUALIZATSIYA")
print("=" * 29)

# 1. Bo'limlar bo'yicha xodimlar statistikasi
dept_analysis = pd.read_sql_query("""
SELECT 
    department,
    COUNT(*) as xodimlar_soni,
    AVG(salary) as ortacha_maosh,
    AVG(age) as ortacha_yosh
FROM employees
GROUP BY department
ORDER BY xodimlar_soni DESC
""", employees_conn)

# 2. Kategoriyalar bo'yicha mahsulotlar
category_analysis = pd.read_sql_query("""
SELECT 
    category,
    COUNT(*) as mahsulotlar_soni,
    AVG(price) as ortacha_narx,
    AVG(rating) as ortacha_reyting
FROM products
GROUP BY category
ORDER BY mahsulotlar_soni DESC
""", products_conn)

# Grafiklar yaratish
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Bo'limlar bo'yicha xodimlar soni
axes[0,0].bar(dept_analysis['department'], dept_analysis['xodimlar_soni'], 
              color='skyblue', alpha=0.8)
axes[0,0].set_title('Bo\'limlar bo\'yicha xodimlar soni', fontweight='bold')
axes[0,0].tick_params(axis='x', rotation=45)
axes[0,0].grid(True, alpha=0.3)

# 2. Bo'limlar bo'yicha o'rtacha maosh
axes[0,1].bar(dept_analysis['department'], dept_analysis['ortacha_maosh'], 
              color='lightcoral', alpha=0.8)
axes[0,1].set_title('Bo\'limlar bo\'yicha o\'rtacha maosh', fontweight='bold')
axes[0,1].tick_params(axis='x', rotation=45)
axes[0,1].grid(True, alpha=0.3)
axes[0,1].yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x/1000:.0f}K'))

# 3. Kategoriyalar bo'yicha mahsulotlar soni
axes[1,0].bar(category_analysis['category'], category_analysis['mahsulotlar_soni'], 
              color='lightgreen', alpha=0.8)
axes[1,0].set_title('Kategoriyalar bo\'yicha mahsulotlar soni', fontweight='bold')
axes[1,0].tick_params(axis='x', rotation=45)
axes[1,0].grid(True, alpha=0.3)

# 4. Kategoriyalar bo'yicha o'rtacha reyting
axes[1,1].bar(category_analysis['category'], category_analysis['ortacha_reyting'], 
              color='orange', alpha=0.8)
axes[1,1].set_title('Kategoriyalar bo\'yicha o\'rtacha reyting', fontweight='bold')
axes[1,1].tick_params(axis='x', rotation=45)
axes[1,1].grid(True, alpha=0.3)
axes[1,1].set_ylim(0, 5)

plt.tight_layout()
plt.show()

print("\n✅ Grafiklar muvaffaqiyatli yaratildi!")
print("📊 SQL so'rovlari natijalarini vizual tahlil qildik")
```
