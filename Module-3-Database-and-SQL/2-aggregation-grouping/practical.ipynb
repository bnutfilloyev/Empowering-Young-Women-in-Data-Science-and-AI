```python
# TODO: Quyidagi so'rovlarni yozing va natijalarni tahlil qiling

# 1. Jami xodimlar sonini toping
query1 = """
-- Sizning javobingiz:

"""

# Javob:
query1_solution = """
SELECT COUNT(*) as total_employees
FROM employees;
"""

df1 = pd.read_sql(query1_solution, engine)
print("👥 Jami xodimlar soni:", df1['total_employees'][0])
```

```python
# 2. Faol mijozlar sonini toping (is_active = true)
query2 = """
-- Sizning javobingiz:

"""

# Javob:
query2_solution = """
SELECT 
    COUNT(*) as total_customers,
    COUNT(CASE WHEN is_active = true THEN 1 END) as active_customers,
    COUNT(CASE WHEN is_active = false THEN 1 END) as inactive_customers
FROM customers;
"""

df2 = pd.read_sql(query2_solution, engine)
print("👥 Mijozlar statistikasi:")
print(f"  Jami: {df2['total_customers'][0]:,}")
print(f"  Faol: {df2['active_customers'][0]:,}")
print(f"  Nofaol: {df2['inactive_customers'][0]:,}")

# Vizualizatsiya
plt.figure(figsize=(8, 6))
labels = ['Faol', 'Nofaol']
sizes = [df2['active_customers'][0], df2['inactive_customers'][0]]
colors = ['lightgreen', 'lightcoral']

plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Mijozlar holati')
plt.axis('equal')
plt.show()
```

### 📈 Vazifa 2: Murakkab COUNT operatsiyalari

```python
# 3. Bo'limlar bo'yicha xodimlar sonini toping va eng ko'p xodimi bo'lgan bo'limni aniqlang
query3 = """
-- Sizning javobingiz:

"""

# Javob:
query3_solution = """
SELECT 
    d.department_name,
    COUNT(e.employee_id) as employee_count,
    ROUND(COUNT(e.employee_id) * 100.0 / (SELECT COUNT(*) FROM employees), 2) as percentage
FROM departments d
LEFT JOIN employees e ON d.department_id = e.department_id
GROUP BY d.department_name, d.department_id
ORDER BY employee_count DESC;
"""

df3 = pd.read_sql(query3_solution, engine)
print("📊 Bo'limlar bo'yicha xodimlar taqsimoti:")
print(df3.to_string(index=False))

# Bar chart
plt.figure(figsize=(12, 6))
plt.bar(df3['department_name'], df3['employee_count'])
plt.xticks(rotation=45, ha='right')
plt.title('Bo\'limlar bo\'yicha xodimlar soni')
plt.xlabel('Bo\'lim nomi')
plt.ylabel('Xodimlar soni')
plt.tight_layout()
plt.show()
```

## 3️⃣ SUM va AVG Funksiyalari bilan Amaliy Mashqlar

### 💰 Vazifa 3: Maosh statistikasi

```python
# 4. Bo'limlar bo'yicha maosh statistikasini hisoblang
query4 = """
-- Sizning javobingiz:
-- Quyidagilarni hisoblang:
-- - Har bir bo'limdagi jami maosh fondi
-- - O'rtacha maosh
-- - Minimal va maksimal maosh
-- - Xodimlar soni

"""

# Javob:
query4_solution = """
SELECT 
    d.department_name,
    COUNT(e.employee_id) as employee_count,
    SUM(e.salary) as total_salary_fund,
    ROUND(AVG(e.salary), 2) as avg_salary,
    MIN(e.salary) as min_salary,
    MAX(e.salary) as max_salary,
    MAX(e.salary) - MIN(e.salary) as salary_range
FROM departments d
LEFT JOIN employees e ON d.department_id = e.department_id
WHERE e.salary IS NOT NULL
GROUP BY d.department_name
ORDER BY avg_salary DESC;
"""

df4 = pd.read_sql(query4_solution, engine)
print("💰 Bo'limlar bo'yicha maosh statistikasi:")
print(df4.to_string(index=False))

# Vizualizatsiya
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# O'rtacha maosh
ax1.bar(df4['department_name'], df4['avg_salary']/1000)
ax1.set_title('Bo\'limlar bo\'yicha o\'rtacha maosh')
ax1.set_ylabel('Maosh (ming so\'m)')
ax1.tick_params(axis='x', rotation=45)

# Maosh diapazoni
ax2.bar(df4['department_name'], df4['salary_range']/1000, color='orange')
ax2.set_title('Bo\'limlar bo\'yicha maosh diapazoni')
ax2.set_ylabel('Maosh farqi (ming so\'m)')
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

### 📊 Vazifa 4: Mahsulot kategoriyalari statistikasi

```python
# 5. Mahsulot kategoriyalari bo'yicha quyidagi ko'rsatkichlarni hisoblang:
query5 = """
-- Sizning javobingiz:
-- - Har bir kategoriyada nechta mahsulot bor
-- - O'rtacha narx
-- - O'rtacha reyting
-- - Jami ombordagi mahsulotlar soni

"""

# Javob:
query5_solution = """
SELECT 
    c.category_name,
    COUNT(p.product_id) as products_count,
    ROUND(AVG(p.price), 2) as avg_price,
    ROUND(AVG(p.rating), 2) as avg_rating,
    SUM(p.stock_quantity) as total_stock,
    MIN(p.price) as min_price,
    MAX(p.price) as max_price
FROM categories c
LEFT JOIN products p ON c.category_id = p.category_id
GROUP BY c.category_name
ORDER BY products_count DESC;
"""

df5 = pd.read_sql(query5_solution, engine)
print("🛍️ Kategoriyalar bo'yicha mahsulotlar statistikasi:")
print(df5.to_string(index=False))

# Scatter plot: O'rtacha narx vs O'rtacha reyting
plt.figure(figsize=(10, 6))
scatter = plt.scatter(df5['avg_price'], df5['avg_rating'], 
                     s=df5['products_count']*10, alpha=0.6, c=df5['total_stock'], 
                     cmap='viridis')

# Kategoriya nomlarini qo'shish
for i, row in df5.iterrows():
    plt.annotate(row['category_name'], 
                 (row['avg_price'], row['avg_rating']),
                 xytext=(5, 5), textcoords='offset points', fontsize=8)

plt.xlabel('O\'rtacha narx (so\'m)')
plt.ylabel('O\'rtacha reyting')
plt.title('Kategoriyalar: Narx vs Reyting\n(Doira hajmi = mahsulotlar soni, rang = ombor)')
plt.colorbar(scatter, label='Jami ombor')
plt.tight_layout()
plt.show()
```

## 4️⃣ GROUP BY va HAVING bilan Amaliy Mashqlar

### 🎯 Vazifa 5: Sotuvlar tahlili

```python
# 6. Hududlar bo'yicha sotuvlar statistikasini hisoblang va faqat 
#    100 mln so'mdan ko'p savdo qilgan hududlarni ko'rsating
query6 = """
-- Sizning javobingiz:
-- GROUP BY va HAVING dan foydalaning

"""

# Javob:
query6_solution = """
SELECT 
    o.delivery_region as region,
    COUNT(DISTINCT o.order_id) as orders_count,
    COUNT(DISTINCT o.customer_id) as unique_customers,
    SUM(o.total_amount) as total_revenue,
    ROUND(AVG(o.total_amount), 2) as avg_order_value,
    ROUND(SUM(o.total_amount)/COUNT(DISTINCT o.customer_id), 2) as revenue_per_customer
FROM orders o
GROUP BY o.delivery_region
HAVING SUM(o.total_amount) > 100000000  -- 100 mln so'mdan ko'p
ORDER BY total_revenue DESC;
"""

df6 = pd.read_sql(query6_solution, engine)
print("🌍 Yuqori savdoli hududlar (100+ mln so'm):")
print(df6.to_string(index=False))

# Bubble chart
plt.figure(figsize=(12, 8))
scatter = plt.scatter(df6['orders_count'], df6['total_revenue']/1e6, 
                     s=df6['unique_customers']/10, alpha=0.6)

for i, row in df6.iterrows():
    plt.annotate(row['region'], 
                 (row['orders_count'], row['total_revenue']/1e6),
                 xytext=(5, 5), textcoords='offset points')

plt.xlabel('Buyurtmalar soni')
plt.ylabel('Jami savdo (mln so\'m)')
plt.title('Yuqori savdoli hududlar\n(Doira hajmi = noyob mijozlar soni)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 📈 Vazifa 6: Murakkab guruhlash

```python
# 7. Mahsulot kategoriyasi va mijoz turi bo'yicha sotuvlar tahlili
#    Faqat 10+ ta sotuv bo'lgan guruhlarni ko'rsating
query7 = """
-- Sizning javobingiz:
-- Ko'p ustun bo'yicha GROUP BY
-- HAVING bilan filtrlash
-- JOIN operatorlaridan foydalaning

"""

# Javob:
query7_solution = """
SELECT 
    c.category_name,
    cust.customer_type,
    COUNT(DISTINCT s.sale_id) as sales_count,
    SUM(s.total_price) as total_revenue,
    ROUND(AVG(s.total_price), 2) as avg_sale_price,
    SUM(s.profit_margin) as total_profit,
    ROUND(SUM(s.profit_margin)/SUM(s.total_price) * 100, 2) as profit_percentage
FROM sales s
JOIN products p ON s.product_id = p.product_id
JOIN categories c ON p.category_id = c.category_id
JOIN orders o ON s.order_id = o.order_id
JOIN customers cust ON o.customer_id = cust.customer_id
GROUP BY c.category_name, cust.customer_type
HAVING COUNT(DISTINCT s.sale_id) >= 10
ORDER BY total_revenue DESC;
"""

df7 = pd.read_sql(query7_solution, engine)
print("📊 Kategoriya va mijoz turi bo'yicha sotuvlar (10+ sotuv):")
print(df7.to_string(index=False))

# Pivot table
pivot_revenue = df7.pivot(index='category_name', 
                         columns='customer_type', 
                         values='total_revenue')

plt.figure(figsize=(12, 8))
sns.heatmap(pivot_revenue/1e6, annot=True, fmt='.1f', cmap='YlOrRd')
plt.title('Kategoriya va mijoz turi bo\'yicha sotuvlar (mln so\'m)')
plt.tight_layout()
plt.show()
```

## 5️⃣ Murakkab Amaliy Vazifalar

### 🏆 Vazifa 7: TOP bo'limlar tahlili

```python
# 8. Quyidagi talablarga javob beradigan so'rov yozing:
#    - Bo'limlar bo'yicha umumiy ko'rsatkichlar
#    - Faqat 5+ xodimi bo'lgan bo'limlar
#    - Xodim boshiga o'rtacha savdo
#    - Bo'lim samaradorligi (savdo/maosh nisbati)

query8 = """
-- Sizning javobingiz:

"""

# Javob:
query8_solution = """
WITH department_stats AS (
    SELECT 
        d.department_name,
        COUNT(DISTINCT e.employee_id) as employee_count,
        SUM(e.salary) as total_salary_cost,
        COUNT(DISTINCT o.order_id) as orders_handled,
        SUM(o.total_amount) as total_sales,
        ROUND(AVG(e.age), 1) as avg_employee_age
    FROM departments d
    LEFT JOIN employees e ON d.department_id = e.department_id
    LEFT JOIN orders o ON e.employee_id = o.employee_id
    GROUP BY d.department_name
    HAVING COUNT(DISTINCT e.employee_id) >= 5
)
SELECT 
    department_name,
    employee_count,
    total_salary_cost,
    orders_handled,
    total_sales,
    ROUND(total_sales/employee_count, 2) as sales_per_employee,
    ROUND(total_sales/total_salary_cost, 2) as efficiency_ratio,
    avg_employee_age
FROM department_stats
ORDER BY efficiency_ratio DESC;
"""

df8 = pd.read_sql(query8_solution, engine)
print("🏆 Bo'limlar samaradorligi tahlili (5+ xodim):")
print(df8.to_string(index=False))

# Multi-metric visualization
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 12))

# Xodim boshiga savdo
ax1.bar(df8['department_name'], df8['sales_per_employee']/1e6)
ax1.set_title('Xodim boshiga savdo')
ax1.set_ylabel('Mln so\'m')
ax1.tick_params(axis='x', rotation=45)

# Samaradorlik koeffitsienti
ax2.bar(df8['department_name'], df8['efficiency_ratio'], color='orange')
ax2.set_title('Samaradorlik (savdo/maosh)')
ax2.set_ylabel('Koeffitsient')
ax2.tick_params(axis='x', rotation=45)

# Buyurtmalar soni
ax3.bar(df8['department_name'], df8['orders_handled'], color='green')
ax3.set_title('Qayta ishlangan buyurtmalar')
ax3.set_ylabel('Buyurtmalar soni')
ax3.tick_params(axis='x', rotation=45)

# O'rtacha yosh
ax4.bar(df8['department_name'], df8['avg_employee_age'], color='red')
ax4.set_title('O\'rtacha xodim yoshi')
ax4.set_ylabel('Yosh')
ax4.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

### 🎯 Vazifa 8: Oylik tendentsiyalar

```python
# 9. Oylik sotuvlar dinamikasini tahlil qiling:
#    - Har oy bo'yicha umumiy sotuvlar
#    - O'sish sur'ati (oldingi oyga nisbatan)
#    - Eng yaxshi va eng yomon oylarni aniqlash

query9 = """
-- Sizning javobingiz:
-- Window functions va DATE functions dan foydalaning

"""

# Javob:
query9_solution = """
WITH monthly_sales AS (
    SELECT 
        TO_CHAR(order_date, 'YYYY-MM') as month,
        COUNT(DISTINCT order_id) as orders_count,
        SUM(total_amount) as monthly_revenue,
        COUNT(DISTINCT customer_id) as unique_customers,
        ROUND(AVG(total_amount), 2) as avg_order_value
    FROM orders
    GROUP BY TO_CHAR(order_date, 'YYYY-MM')
    ORDER BY month
)
SELECT 
    month,
    orders_count,
    monthly_revenue,
    unique_customers,
    avg_order_value,
    LAG(monthly_revenue) OVER (ORDER BY month) as prev_month_revenue,
    ROUND(
        (monthly_revenue - LAG(monthly_revenue) OVER (ORDER BY month)) * 100.0 / 
        LAG(monthly_revenue) OVER (ORDER BY month), 2
    ) as growth_percentage
FROM monthly_sales
ORDER BY month;
"""

df9 = pd.read_sql(query9_solution, engine)
print("📈 Oylik sotuvlar dinamikasi:")
print(df9.to_string(index=False))

# Time series visualization
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# Oylik daromad
ax1.plot(df9['month'], df9['monthly_revenue']/1e6, marker='o', linewidth=2)
ax1.set_title('Oylik daromad dinamikasi')
ax1.set_ylabel('Daromad (mln so\'m)')
ax1.grid(True, alpha=0.3)
ax1.tick_params(axis='x', rotation=45)

# O'sish sur'ati
growth_data = df9[df9['growth_percentage'].notna()]
colors = ['green' if x >= 0 else 'red' for x in growth_data['growth_percentage']]
ax2.bar(growth_data['month'], growth_data['growth_percentage'], color=colors)
ax2.set_title('Oylik o\'sish sur\'ati (%)')
ax2.set_ylabel('O\'sish (%)')
ax2.axhline(y=0, color='black', linestyle='-', alpha=0.3)
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Statistik xulosa
print("\n📊 Oylik tendentsiyalar xulosasi:")
print(f"Eng yuqori daromad: {df9['monthly_revenue'].max()/1e6:.1f} mln so'm ({df9.loc[df9['monthly_revenue'].idxmax(), 'month']})")
print(f"Eng past daromad: {df9['monthly_revenue'].min()/1e6:.1f} mln so'm ({df9.loc[df9['monthly_revenue'].idxmin(), 'month']})")
if not df9['growth_percentage'].isna().all():
    print(f"Eng yuqori o'sish: {df9['growth_percentage'].max():.1f}% ({growth_data.loc[growth_data['growth_percentage'].idxmax(), 'month']})")
    print(f"Eng katta pasayish: {df9['growth_percentage'].min():.1f}% ({growth_data.loc[growth_data['growth_percentage'].idxmin(), 'month']})")
```

## 6️⃣ Yakuniy Vazifa - Barchani Birlashtirish

### 🏅 Vazifa 9: Kompleks biznes tahlili

```python
# 10. Yakuniy tahlil vazifasi:
#     Kompaniya rahbariyati uchun hisobot tayyorlang:
#     - TOP 5 mijoz (eng ko'p xarid qilgan)
#     - TOP 5 mahsulot (eng ko'p sotilgan)
#     - Eng samarali xodimlar
#     - Hudud bo'yicha tahlil
#     - Tavsiyalar

print("🏅 KOMPLEKS BIZNES TAHLILI")
print("=" * 50)

# TOP 5 mijozlar
top_customers_query = """
SELECT 
    c.customer_id,
    c.first_name || ' ' || c.last_name as customer_name,
    c.region,
    c.customer_type,
    COUNT(DISTINCT o.order_id) as total_orders,
    SUM(o.total_amount) as total_spent,
    ROUND(AVG(o.total_amount), 2) as avg_order_value
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.first_name, c.last_name, c.region, c.customer_type
ORDER BY total_spent DESC
LIMIT 5;
"""

top_customers = pd.read_sql(top_customers_query, engine)
print("\n👑 TOP 5 MIJOZLAR:")
print(top_customers.to_string(index=False))

# TOP 5 mahsulotlar
top_products_query = """
SELECT 
    p.product_name,
    c.category_name,
    COUNT(DISTINCT s.sale_id) as sales_count,
    SUM(s.total_price) as total_revenue,
    ROUND(AVG(p.rating), 2) as avg_rating,
    p.price as unit_price
FROM products p
JOIN categories c ON p.category_id = c.category_id
JOIN sales s ON p.product_id = s.product_id
GROUP BY p.product_id, p.product_name, c.category_name, p.price
ORDER BY total_revenue DESC
LIMIT 5;
"""

top_products = pd.read_sql(top_products_query, engine)
print("\n🏆 TOP 5 MAHSULOTLAR:")
print(top_products.to_string(index=False))

# TOP xodimlar
top_employees_query = """
SELECT 
    e.first_name || ' ' || e.last_name as employee_name,
    d.department_name,
    COUNT(DISTINCT o.order_id) as orders_processed,
    SUM(o.total_amount) as total_sales,
    e.salary,
    ROUND(SUM(o.total_amount)/e.salary, 2) as sales_to_salary_ratio
FROM employees e
JOIN departments d ON e.department_id = d.department_id
JOIN orders o ON e.employee_id = o.employee_id
GROUP BY e.employee_id, e.first_name, e.last_name, d.department_name, e.salary
HAVING COUNT(DISTINCT o.order_id) >= 5
ORDER BY sales_to_salary_ratio DESC
LIMIT 5;
"""

top_employees = pd.read_sql(top_employees_query, engine)
print("\n⭐ ENG SAMARALI XODIMLAR:")
print(top_employees.to_string(index=False))
```

```python
# Vizualizatsiya - Dashboard
fig = plt.figure(figsize=(16, 12))

# TOP mijozlar
ax1 = plt.subplot(2, 3, 1)
plt.barh(top_customers['customer_name'], top_customers['total_spent']/1e6)
plt.title('TOP 5 Mijozlar')
plt.xlabel('Jami xarid (mln so\'m)')

# TOP mahsulotlar
ax2 = plt.subplot(2, 3, 2)
plt.bar(range(len(top_products)), top_products['total_revenue']/1e6)
plt.title('TOP 5 Mahsulotlar')
plt.ylabel('Daromad (mln so\'m)')
plt.xticks(range(len(top_products)), 
           [name[:15] + '...' if len(name) > 15 else name for name in top_products['product_name']], 
           rotation=45)

# TOP xodimlar
ax3 = plt.subplot(2, 3, 3)
plt.barh(top_employees['employee_name'], top_employees['sales_to_salary_ratio'])
plt.title('Eng Samarali Xodimlar')
plt.xlabel('Savdo/Maosh nisbati')

# Hududlar bo'yicha tahlil
ax4 = plt.subplot(2, 3, 4)
region_stats = pd.read_sql("""
    SELECT delivery_region, SUM(total_amount) as revenue
    FROM orders GROUP BY delivery_region
    ORDER BY revenue DESC LIMIT 10
""", engine)
plt.pie(region_stats['revenue'], labels=region_stats['delivery_region'], autopct='%1.1f%%')
plt.title('Hududlar bo\'yicha daromad')

# Kategoriyalar bo'yicha sotuvlar
ax5 = plt.subplot(2, 3, 5)
category_stats = pd.read_sql("""
    SELECT c.category_name, COUNT(s.sale_id) as sales
    FROM categories c 
    LEFT JOIN products p ON c.category_id = p.category_id
    LEFT JOIN sales s ON p.product_id = s.product_id
    GROUP BY c.category_name
    ORDER BY sales DESC
""", engine)
plt.bar(category_stats['category_name'], category_stats['sales'])
plt.title('Kategoriyalar bo\'yicha sotuvlar')
plt.xticks(rotation=45)

# Umumiy ko'rsatkichlar
ax6 = plt.subplot(2, 3, 6)
summary_stats = pd.read_sql("""
    SELECT 
        COUNT(DISTINCT customer_id) as customers,
        COUNT(DISTINCT product_id) as products,
        COUNT(DISTINCT employee_id) as employees,
        COUNT(*) as total_orders,
        SUM(total_amount) as total_revenue
    FROM orders o
    JOIN sales s ON o.order_id = s.order_id
    JOIN employees e ON o.employee_id = e.employee_id
""", engine)

metrics = ['Mijozlar', 'Mahsulotlar', 'Xodimlar', 'Buyurtmalar']
values = [summary_stats['customers'][0], summary_stats['products'][0], 
          summary_stats['employees'][0], summary_stats['total_orders'][0]]
plt.bar(metrics, values)
plt.title('Umumiy ko\'rsatkichlar')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

print(f"\n💰 UMUMIY NATIJALAR:")
print(f"  📊 Jami daromad: {summary_stats['total_revenue'][0]/1e9:.2f} milliard so'm")
print(f"  👥 Faol mijozlar: {summary_stats['customers'][0]:,}")
print(f"  🛍️ Mahsulotlar assortimenti: {summary_stats['products'][0]:,}")
print(f"  👨‍💼 Xodimlar soni: {summary_stats['employees'][0]:,}")
print(f"  📦 Jami buyurtmalar: {summary_stats['total_orders'][0]:,}")
```
