## 2️⃣ Maosh va Kompensatsiya Tahlili (25 ball)

### 🎯 Vazifa: Maosh tizimining adalatliligini baholash

**Talablar:**
- Bo'limlar orasida maosh farqlari
- Yosh va tajribaga qarab maosh o'zgarishi
- Gender pay gap tahlili
- Maosh diapazoni va outlierlar

```python
# 💰 Maosh adalatligi tahlili
# TODO: Guruh a'zolari maosh tizimini chuqur tahlil qilsin

salary_fairness_query = """
-- Guruh 2 ning maosh tahlili:
-- Bo'limlar va yosh guruhlar bo'yicha:
-- - O'rtacha maosh
-- - Median maosh
-- - Standart og'ish
-- - 25% va 75% percentile
-- - Eng katta va eng kichik maosh
-- - Maosh teng taqsimlanganmi?


"""

salary_analysis = pd.read_sql(salary_fairness_query, engine)
print("💰 MAOSH ADALATLIGI TAHLILI:")
print(salary_analysis.to_string(index=False))
```

```python
# 📊 Maosh outlierlarini topish
# TODO: Guruh a'zolari anomal maoshlarni aniqlsin

salary_outliers_query = """
-- Guruh vazifasi: Maosh outlierlarini topish
-- Quyidagi xodimlarni aniqlang:
-- - Bo'limidagi o'rtachadan 2 marta ko'p olganlar
-- - Bo'limidagi o'rtachadan 2 marta kam olganlar
-- - Yosh guruhidagi o'rtachadan juda farq qiladiganlar
-- - Mumkin bo'lgan sabablari


"""

salary_outliers = pd.read_sql(salary_outliers_query, engine)
print("🚨 MAOSH ANOMALIYALARI:")
print(salary_outliers.to_string(index=False))
```

```python
# 📈 Maosh ko'rsatkichlari vizualizatsiyasi
# TODO: Guruh a'zolari maosh taqsimotini vizuallashtirsin

# Box plot: Bo'limlar bo'yicha maosh taqsimoti
salary_distribution_query = """
SELECT d.department_name, e.salary
FROM employees e
JOIN departments d ON e.department_id = d.department_id
WHERE e.salary IS NOT NULL;
"""

salary_dist = pd.read_sql(salary_distribution_query, engine)

plt.figure(figsize=(12, 8))
sns.boxplot(x='department_name', y='salary', data=salary_dist)
plt.xticks(rotation=45)
plt.title('Bo\'limlar bo\'yicha maosh taqsimoti')
plt.ylabel('Maosh (so\'m)')
plt.tight_layout()
plt.show()

# Guruh vazifasi: Qo'shimcha grafiklar yarating
```
